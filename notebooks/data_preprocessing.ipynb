# Data Preprocessing for Network Anomaly Detection

This notebook contains the code for preprocessing the raw data used for network anomaly detection. The preprocessing steps include data cleaning, normalization, and feature extraction.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Load raw data
def load_data(file_path):
    data = pd.read_csv(file_path)
    return data

# Data cleaning function
def clean_data(data):
    # Remove duplicates
    data = data.drop_duplicates()
    # Fill missing values
    data = data.fillna(method='ffill')
    return data

# Normalize features
def normalize_data(data):
    scaler = StandardScaler()
    scaled_data = scaler.fit_transform(data)
    return pd.DataFrame(scaled_data, columns=data.columns)

# Feature extraction example
def feature_extraction(data):
    # Example: Create new features based on existing ones
    data['new_feature'] = data['feature1'] / data['feature2']
    return data

# Main preprocessing function
def preprocess_data(file_path):
    raw_data = load_data(file_path)
    cleaned_data = clean_data(raw_data)
    normalized_data = normalize_data(cleaned_data)
    final_data = feature_extraction(normalized_data)
    return final_data

# Example usage
# processed_data = preprocess_data('data/raw/your_raw_data.csv')
```
